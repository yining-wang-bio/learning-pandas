```python
import pandas as pd
df = pd.read_csv("xuezhi.csv")
```

```python
df
```

```python
#2.布尔索引筛选(找出血脂大于5.5的高危样本)
print("===高危样本(血脂>5.5)===")
high_risk=df[df['血脂']>5.5]
print(high_risk)
print("\n")
```

```python
#3.数据排序(按血脂从高到低排列,看谁最高)
print("===血脂降序排列===")
sorted_df=df.sort_values(by='血脂',ascending=False)
print(sorted_df)
print("\n")
```

```python
#4.描述性统计(一键算出平均值,最大值，最小值等)
print("===血脂描述性统计===")
stats=df['血脂'].describe()
print(stats)
```
