```python
from scipy.stats import chi2_contingency
import pandas as pd

# 颈麻: 有效16 无效10: 置栓: 有效11 无效4
table = pd.DataFrame([[16,10], [11,4]],
index=["颈麻","置栓"], columns=["有效", "无效"])
chi2,p,dof,expected =chi2_contingency(table)
print(f"chi2={chi2:.3f}, P={p:.4f}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif'] = ['SimHei']

data = {"血型": ["A","A","AB","AB","B","B","O","O"],
        "血脂":[5.5,6.0,4.2,4.5,4.4,4.8,6.1,6.3]}
df = pd.DataFrame(data)
df.groupby("血型")["血脂"].mean().plot(kind="bar",title="不同血型患者血脂水平")
plt.show()
```
